# DictVectorizer把类别型数据，变为one-hot编码

```python
from sklearn.feature_extraction import DictVectorizer

data = [
    {'颜色': '红', '尺寸': '大', '价格': 100},
    {'颜色': '蓝', '尺寸': '中', '价格': 80},
    {'颜色': '红', '尺寸': '小', '价格': 50}
]

vec = DictVectorizer(sparse=False)
data_vec = vec.fit_transform(data)

print(vec.get_feature_names_out()) # 获取特征名称
print(data_vec) # 获取特征向量
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
documents = [
    "I love machine learning. Machine learning is interesting.",
    "I love coding. Coding is fun and useful.",
    "Machine learning and coding are my favorite skills."
]

vectorizer = CountVectorizer(min_df=1) #调超参，min_df=1，表示最小文档频率为1，即每个词至少在1个文档中出现
X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out()) # 获取特征名称
print(X.toarray()) # 获取特征向量

# 结合DataFrame查看
df = pd.DataFrame(X.toarray(), columns = vectorizer.get_feature_names_out())
print(df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "人生苦短，我喜欢 python", 
    "人生漫长，不用 python python",
    "人生漫漫，不用 python python python"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())  # 输出特征名称
print(X.toarray())                         # 输出特征向量
```

```python
import jieba
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "人生苦短，我喜欢 python", 
    "人生漫长，不用 python python",
    "人生漫漫，不用 python python python"
]

# 使用jieba分词, 把每个句子分词，然后通过空格再合并成一个句子
# jieba.lcut 和 jieba.cut 的区别如下：
# - jieba.lcut(sentence) 直接返回分词后的列表。
# - jieba.cut(sentence) 返回的是一个可迭代的生成器，需要用 list() 转换成列表才可查看全部分词结果。
documents_cut = [" ".join(jieba.lcut(doc)) for doc in documents]
print(documents_cut)

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents_cut)

print(vectorizer.get_feature_names_out())  # 输出特征名称
print(X.toarray())                         # 输出特征向量
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love machine learning. Machine learning is interesting.",
    "I love coding. Coding is fun and useful.",
    "Machine learning and coding are my favorite skills."
]

vectorizer = TfidfVectorizer()        # 创建一个TF-IDF向量器对象
X = vectorizer.fit_transform(documents)   # 用vectorizer对文本进行拟合，并转化为TF-IDF矩阵

print(vectorizer.get_feature_names_out())   # 输出特征名称
print(X.toarray())                         # 输出TF-IDF特征向量

import pandas as pd
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
print(df)
```
