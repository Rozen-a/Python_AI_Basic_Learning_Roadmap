```python
from sklearn.datasets import load_iris
import pandas as pd

# 加载鸢尾花数据集
iris = load_iris()


# 查看特征名称
print("特征名称:", iris.feature_names)

# 查看目标名称
print("类别名称:", iris.target_names)

# 查看数据集形状
print("数据形状:", iris.data.shape)
print("标签形状:", iris.target.shape)


print(iris.target)

# 将数据转换为DataFrame, 方便查看
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
print(df.head())

df['target_name'] = df['target'].map(lambda x: iris.target_names[x]) #把target列的值映射为类别名称

# 展示前几行
print("数据示例:")
print(df.head())
```

```python
from sklearn.model_selection import train_test_split

# 特征和标签
X = df[iris.feature_names]
y = df['target']

# 划分数据集，默认25%为测试集
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

print("训练集样本数:", X_train.shape[0])
print("测试集样本数:", X_test.shape[0])

# 输出测试集的特征shape和标签shape
print("测试集特征 shape:", X_test.shape)
print("测试集标签 shape:", y_test.shape)

```

```python
type(X_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 创建KNN分类器实例
knn = KNeighborsClassifier(n_neighbors=5)

# 拟合模型
knn.fit(X_train, y_train)

# 预测测试集
y_pred = knn.predict(X_test)

# print(y_pred)
# print(y_test)
# 计算准确率
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
print("KNN准确率: {:.2f}%".format(accuracy * 100))
```
