```python
import numpy as np
import time

# 列表处理1000万个数字的加法（慢，需循环）
start = time.time()
list1 = [i for i in range(10000000)]
list2 = [i*2 for i in range(10000000)]
result = [a + b for a, b in zip(list1, list2)]  # 需循环
end = time.time()
print(f"列表处理1000万个数字的加法耗时：{end-start}")

# NumPy数组处理（快，直接运算）
arr1 = np.arange(10000000)
arr2 = np.arange(10000000) * 2
result = arr1 + arr2  # 无需循环，直接矢量化运算
time3 = time.time()
print(f"NumPy数组处理1000万个数字的加法耗时：{time3-end}")
```

# 4.1 形状改变

```python
# 一维变多维
arr9 = np.arange(0, 12)
arr10 = arr9.reshape(3,4)

# 说明: 不改变原有矩阵，返回改变之后的新矩阵
print(arr9)
print(arr10)
```

```python
# 默认情况下‘C’以行为主的顺序展开，‘F’（Fortran风格）意味着以列的顺序展开,不重要，用不上
arr15 = arr10.reshape((12,),order='F')
print(arr15)
```

```python
arr11=arr10+2
print(arr11)
arr12=arr11.reshape(12)  #再变回一维，元素位置不变
print(arr12)
```

# 4.2 类型转换

# 4.3 索引与切片

```python
# 1. 列表的索引与切片
# list[start:end:step]
arr = [1, 2, 3, 4, 5, 6]
print(arr[0:4:2])           # 1,3

# 2. 一维ndarray的索引与切片， 和列表的索引与切片一样
arr14 = np.arange(0, 12)
print(arr14[0:4:2])         # [0,2]

print('-'*50)
# 3. 多维ndarray的索引与切片
arr15 = np.arange(0, 12).reshape(3,4)
print(arr15)

# 取一行(下标为1的一行, 也就是第二行)
print(arr15[1,:])

# 取一列(下标为1的一列, 也就是第二列)
print(arr15[:,1])

# 取不连续的多行
print(arr15[[0,2],:])

# 取从第2行开始的所有行
print(arr15[1:,:])
```

```python
arr16 = np.arange(0, 24).reshape(2,3,4)
print(arr16)
print('-'*50)
print(arr16[1,:])
print('-'*50)
print(arr16[1,:,:])
print('-'*50)
print(arr16[:,1])
print('-'*50)
print(arr16[:,[0,2],:])
print('-'*50)
print(arr16[:,:,2:])
```

# 4.4 元素的修改

```python
import numpy as np
arr15 = np.arange(0, 12).reshape(3,4)
print(arr15)
arr15[[1,2],[0,3]]
```

```python
# 元素修改
arr15 = np.arange(0, 12).reshape(3,4)

print(arr15)
# 修改单个值
arr15[1,1] = 100
print(arr15)
print('-'*50)
arr15 = np.arange(0, 12).reshape(3,4)
# 修改某一行的值
arr15[1,:] = 100
print(arr15)
print('-'*50)
# 修改某一列的值
arr15[:,1] = 100
print(arr15)
print('-'*50)
arr15 = np.arange(0, 12).reshape(3,4)
# 修改某一个块的值
arr15[0:2,0:2] = 100
print(arr15)
print('-'*50)
# 修改多个不相邻的点 (0,0) (1,1) (2,3)
arr15[[0,1,2],[0,1,3]] = 99
print(arr15)
print('-'*50)
# 根据条件修改
arr15 = np.arange(0, 12).reshape(3,4)
arr15[arr15>5] = 0
print(arr15)
print('-'*50)
# 三目运算
result = np.where(arr15>0,10,20)
print(result)
```

# 5 数组的运算

```python
arr16 = np.arange(0, 12).reshape(3,4)
print(arr16)
print('-' * 20)
print(arr16 + 10) # 加法，有广播机制，每一个元素都加上10
print('-'*20)
print(arr16 - 5)
print('-'*20)
print(arr16 * 2)
print('-'*20)
print(arr16 / 2)
```

```python
# 1. 同种形状的矩阵运算，运算结果是按对应位置 进行运算
arr17 = np.arange(10, 22).reshape(3,4)
print(arr17)
arr18 = np.arange(1, 13).reshape(3,4)
print(arr18)

# 加法
print('-'*50)
print(arr17 + arr18)
# 减法
print('-'*50)
print(arr17 - arr18)
# 乘法
print('-'*50)
print(arr17 * arr18)
# 除法
print('-'*50)
print(arr17 / arr18)
```

```python
#不同形状的，运算出错
arr19 = np.arange(0, 12).reshape(3,4)
arr20 = np.arange(0, 4).reshape(2,2)
# print(arr19 + arr20)
```

# 当两个数组形状不同时，其中一个数组的列数或行数必须为1，广播运算。如果是3维及以上的，依然是这个规律

```python

# 列数相同（行形状相同）
arr21 = np.arange(0, 12).reshape(3,4)       # 3行4列
print(arr21)
arr22 = np.arange(0, 4).reshape(1,4)                
print(arr22)

# 依然是广播的特性
print(arr22-arr21)


print('-'*50)
# 行数相同（列形状相同）
arr23 = np.arange(0, 12).reshape(3,4)
print(arr23)
arr24 = np.arange(0, 3).reshape(3,1)
print(arr24)
# 依然是广播的特性
print(arr23-arr24)
```

```python
#产生2个3维的数组
arr25 = np.arange(0, 24).reshape(2,3,4)
arr26 = np.arange(0, 8).reshape(2,1,4)
result=arr25+arr26
print(result.shape)
```
