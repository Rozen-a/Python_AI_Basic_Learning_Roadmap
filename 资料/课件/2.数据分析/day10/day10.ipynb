# 默写缺失值的处理过程

```python

import pandas as pd
import numpy as np

# 假设有一个示例 DataFrame
data = {
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': [10, 11, 12, 13]
}
df = pd.DataFrame(data)
print(df)

# 1. 查看缺失值情况
print("各列缺失值数量：")
print(df.isnull().sum())

# 2. 删除含有缺失值的行
df_dropna = df.dropna()
print("\n删除含有缺失值的行后：")
print(df_dropna)

# 3. 用指定值填充缺失值
df_fillna = df.fillna(0)
print("\n用0填充缺失值后：")
print(df_fillna)

# 4. 用均值或中位数等统计量填充缺失值
df_fillmean = df.fillna(df.mean())
print("\n用均值填充缺失值后：")
print(df_fillmean)
```

# 完成鸢尾花，20类新闻数据集的加载及理解

```python
from sklearn import datasets
import pandas as pd

# 加载鸢尾花数据集
iris = datasets.load_iris()

# 查看数据集的基本信息
print("特征名称：", iris.feature_names)
print("目标名称：", iris.target_names)

# 构建DataFrame
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target

# 查看前5行数据
print(iris_df.head())

# 查看每一类样本数量
print("\n每一类样本数量：")
print(iris_df['target'].value_counts())

# 描述性统计
print("\n描述性统计：")
print(iris_df.describe())
```

```python
from sklearn.datasets import fetch_20newsgroups

# 加载20类新闻数据集
newsgroups = fetch_20newsgroups(subset='all')

# 查看20类新闻数据集的基本信息
print("类别数：", len(newsgroups.target_names))
print("类别名称：", newsgroups.target_names)
print("新闻样本数量：", len(newsgroups.data))

# 输出前2个文本样本和对应类别
for i in range(2):
    print(f"\n第{i+1}条新闻样本文本：\n", newsgroups.data[i][:300], "...")  # 打印前300字符
    print("对应的类别：", newsgroups.target_names[newsgroups.target[i]])
```

# 机器学习的DictVectorizer练习，CountVectorizer, TfidfVectorizer练习

```python
from sklearn.feature_extraction import DictVectorizer

data = [
    {'颜色': '红', '尺寸': '大', '价格': 100},
    {'颜色': '蓝', '尺寸': '中', '价格': 80},
    {'颜色': '红', '尺寸': '小', '价格': 50}
]
vcs=DictVectorizer()
vcs.fit_transform(data).toarray()
data_vcs=vcs.fit_transform(data).toarray()
print(vcs.get_feature_names_out())
print(data_vcs)


from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love machine learning. Machine learning is interesting.",
    "I love coding. Coding is fun and useful.",
    "Machine learning and coding are my favorite skills."
]
# 创建CountVectorizer对象
cv=CountVectorizer()
cv.fit_transform(documents).toarray()
data_cv=cv.fit_transform(documents).toarray()
print(cv.get_feature_names_out())
print(data_cv)

from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "人生苦短，我喜欢 python", 
    "人生漫长，不用 python python",
    "人生漫漫，不用 python python python"
]
# 创建TfidfVectorizer对象
tf = TfidfVectorizer()
tf.fit_transform(documents).toarray()
data_tf=tf.fit_transform(documents).toarray()
print(tf.get_feature_names_out())
print(data_tf)




```
