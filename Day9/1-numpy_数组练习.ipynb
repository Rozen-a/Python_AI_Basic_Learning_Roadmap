**4、完成数组的添加练习**

```python
arr1 = np.arange(0, 6).reshape(2, 3)
print('原数组：')
print(arr1)

# 默认展开为一维数组再添加
arr2 = np.append(arr1, np.array([7, 8, 9]))
print('默认添加：')
print(arr2)

# 指定轴添加
# 沿0轴添加，1轴需与arr1相同
arr3 = np.append(arr1, np.array([[7, 8, 9]]), axis=0)
print('指定轴添加：')
print(arr3)

# 沿1轴添加，0轴需与arr1相同
arr4 = np.append(arr1, np.array([[7], [8]]), axis=1)
print('沿1轴添加：')
print(arr4)
```

**5、完成数组的连接，堆叠**

5.1 数组的连接

```python
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([[7,8,9],[10,11,12]])

# 默认演0轴拼接
arr3 = np.concatenate((arr1, arr2))
print(arr3)

# 沿1轴拼接
arr4 = np.concatenate((arr1, arr2), axis=1)
print(arr4)
```

5.2 数组堆叠

```python
# 一维数组堆叠
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr3 = np.array([7, 8, 9])

# 按0轴堆叠
print(np.stack([arr1, arr2, arr3]))

# 按1轴堆叠
print(np.stack([arr1, arr2, arr3], axis=-1))
```

```python
# 二维数组堆叠
arr1 = np.arange(12).reshape(3, 4)
arr2 = np.arange(12, 24).reshape(3, 4)

# 沿0轴堆叠
arr3 = np.stack([arr1, arr2], axis=0)
print(arr3)
print(arr3.shape)

# 沿1轴堆叠
arr4 = np.stack([arr1, arr2], axis=1)
print(arr4)
print(arr4.shape)
```

**6、完成nan和inf的使用,与上课一致**

```python
# 创建一个nan和inf
a= np.nan  # 非数
b = np.inf  # 正无穷
c = -np.inf  # 负无穷

# 都是float类型
print(a, type(a))
print(b, type(b))
print(c, type(c))

print(b > c)  # 正无穷比负无穷大

print(np.inf == np.inf)  # 正无穷等于正无穷
print(np.nan == np.nan)  # 非数不等于非数
```

```python
arr = np.arange(24, dtype=float).reshape(4, 6)
print(arr)

# 将指定位置的数给成nan
arr[3, 4] = np.nan
print(arr)

# 判断非零的个数
print(np.count_nonzero(arr))

# 判断nan（缺失值）个数
print(np.count_nonzero(arr != arr))  # 只有nan != nan 为Ture， 其余都是False

# 将 nan 替换为 0
arr[np.isnan(arr)] = 0
print(arr)
```

**7、np二维数组转置练习，轴滚动练习**

```python
arr = np.arange(0, 12).reshape(3, 4)
print('原数组：')
print(arr)

# 转置的第一种方法
arr1 = arr.T
print('第一种方法装置后的数组：')
print(arr1)

# 转置的第二种方法
arr2 = np.transpose(arr)
print('第二种方法装置后的数组：')
print(arr2)

# 对换数组的轴
arr3 = np.transpose(arr)
print('对换数组的轴后的数组：')
print(arr3)
```

```python
arr = np.arange(0, 12).reshape(1, 3, 4)
print(arr.shape)

# 三维数组对换轴，指定对换后轴顺序
arr1 = np.transpose(arr, (2, 0, 1))
print(arr1.shape)
```

**11、练习重复值，缺失值处理**

11.1 检测重复值

```python
data = {
    "name": ["Alice", "Alice", "Charlie", "David", "Alice", "Charlie"],
    "age": [20, 22, 21, 20, 20, 21],
    "score": [85, 90, 78, 92, 85, 78],
    "gender": ["女", "男", "男", "男", "女", "男"]
}
df = pd.DataFrame(data, index=['s1', 's2', 's3', 's4', 's5', 's6'], columns=['name', 'age', 'score', 'gender'])
print(df)
print('-'*20)

# 1. 检测重复值
print(df.duplicated())      # 返回布尔Series，True表示重复行
print('-'*20)

# 2. 指定列检测重复值(可以指定多列)
# 参数 keep, 默认值为'first', 表示只保留第一个重复行, 'last'表示只保留最后一个重复行, False表示标记所有重复行
df_dup_by_col = df.duplicated(subset=['age'], keep='first')
print(df_dup_by_col)
```

11.2 删除重复值

```python
# 创建一个数据表
data = {
    "name": ["Alice", "Alice", "Charlie", "David", "Alice", "Charlie"],
    "age": [20, 22, 21, 20, 20, 21],
    "score": [85, 90, 78, 92, 85, 78],
    "gender": ["女", "男", "男", "男", "女", "男"]
}
df = pd.DataFrame(data, index=['s1', 's2', 's3', 's4', 's5', 's6'], columns=['name', 'age', 'score', 'gender'])
print("原数据:")
print(df)

# 1. 删除完全重复的行
df_drop_dup = df.drop_duplicates()
print("删除完全重复的行:")
print(df_drop_dup)

# 2. 删除部分列重复的行
df_drop_dup_by_col = df.drop_duplicates(subset=['name'], keep='first')
print("删除部分列重复的行:")
print(df_drop_dup_by_col)

# 3. 直接在原数据上删除
df.drop_duplicates(inplace=True)
print("直接在原数据上删除:")
print(df)
```

11.4 删除缺失值

11.5 填充缺失值

```python
#填充缺失值
# 1. 创建一个数据表
data = {
    "name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "age": [20, 22, 21, 20, 20, np.nan],
    "score": [85, 90, np.nan, 92, 85, np.nan],
    "gender": ["女", "男", "男", np.nan, np.nan,np.nan]
}
df = pd.DataFrame(data, index=['s1', 's2', 's3', 's4', 's5', 's6'], columns=['name', 'age', 'score', 'gender'])
print("原数据:")
print(df)

# 2. 填充全局缺失值
df_fill_na = df.fillna(0)
print("填充缺失值为0:")
print(df_fill_na)

# 3. 填充某一列的缺失值
df_fill_na_by_col = df.fillna({'gender': '男','score':80})
print("填充【gender】缺失值为男，【score】缺失值为80:")
print(df_fill_na_by_col)


# 4.单独把某列取出来，填充缺失值
df['score'] = df['score'].fillna(80)
print("单独把【score】列取出来，填充缺失值为80:")
print(df)
```
