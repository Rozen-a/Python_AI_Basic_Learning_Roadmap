**1、默写缺失值的处理过程**

```python
import numpy as np
import pandas as pd
from pandas.io.formats import printing
# 创建数据
data = {
    'name':['张三','李四','王五','赵六', '小明', '小红'],
    'age':[20,21,22,23, np.nan, 25],
    'sex':['男','女','男','女','男',np.nan],
    'score':[95,85,np.nan,60,np.nan,78]
}
df = pd.DataFrame(data)
print('原数据：')
print(df)

# 查看各列缺失值个数
print('\n各列缺失值个数：')
print(df.isnull().sum())

# 删除非数值列包含缺失值的行
df_dropna = df.dropna(subset=['name', 'sex'])
print('\n删除非数值列包含缺失值的行：')
print(df_dropna)

# 用均值填充缺失值
df_fillna = df_dropna.fillna({
    'age':df_dropna['age'].mean(),  # 计算 age 列的平均值，用这个平均值去填补 age 列中所有的缺失值
    'score':df_dropna['score'].mean()  # 计算 score 列的平均值，用这个平均值去填补 score 列中所有的缺失值
})
print('\n用均值填充数值列缺失值：')
print(df_fillna)
```

**2、完成鸢尾花，20类新闻数据集的加载及理解**

2.1 鸢尾花示例

```python
# 导入数据集
from sklearn.datasets import load_iris

# 数据集加载
data = load_iris()

# 提取特征值和目标值
X = data.data
y = data.target

# 查看数据基本信息
print('特征名称：', data.feature_names)
print('目标名称：', data.target_names)
print(f'数据集规模：特征{X.shape}，目标{y.shape}')
```

```python
# 数据预处理
from sklearn.model_selection import train_test_split  # 数据划分
from sklearn.preprocessing import StandardScaler  # 标准化

# 数据集划分
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print('训练集规模：', X_train.shape)

# 标准化
scaler = StandardScaler()  # 特征标准化对象
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# 模型训练
from sklearn.neighbors import KNeighborsClassifier

# 初始化KNN分类器
knn = KNeighborsClassifier(n_neighbors=5)

# 训练模型
knn.fit(X_train, y_train)
```

```python
# 模型预测与评估
from sklearn.metrics import accuracy_score

# 模型预测
y_pred = knn.predict(X_test)

# 评估
print(f'模型准确率：{accuracy_score(y_test, y_pred):.2f}')
```

2.2 20类新闻数据集示例

```python
from sklearn.datasets import fetch_20newsgroups

# 1. 加载 20 类新闻数据集（加载全部数据）
news = fetch_20newsgroups(data_home='./data', subset='all', shuffle=True, random_state=42)
'''
data_home='./data'：
指定数据在本地保存的目录。
 - 如果 ./data 中还没有该数据集，函数会自动去网上下载到这个目录。
 - 下次再调用时，会从本地 ./data 直接读取，不再重新下载

subset='all'：
指定加载数据集的哪一部分：
 - 'train'：只加载训练集
 - 'test'：只加载测试集
 - 'all'：训练集 + 测试集全部加载（你这里就是这种情况）

shuffle=True：
 - 在加载数据后，打乱样本顺序。

random_state=42：
 - 设置随机数种子，使打乱顺序的结果可复现。
'''
# 2. 查看数据基本结构
print("keys:", news.keys())
print("样本数:", len(news.data))
print("标签数:", len(news.target))
print("类别数:", len(news.target_names))
print("类别列表:", news.target_names)
print("文本示例：\n", news.data[0])
print("该文本的分类标签:", news.target[0], "-", news.target_names[news.target[0]])
```

**3、机器学习的DictVectorizer练习，CountVectorizer, TfidfVectorizer练习**

3.1 从字典加载特征

```python
# 导入字典提取器
from sklearn.feature_extraction import DictVectorizer

# 字典数据
data = [
    {'颜色': '红', '尺寸': '大', '价格': 100},
    {'颜色': '蓝', '尺寸': '中', '价格': 80},
    {'颜色': '红', '尺寸': '小', '价格': 50}
]

# 初始化字典提取器
# sparse=False 表示返回的矩阵是二维的，而不是稀疏矩阵
# sparse=True 默认值，返回的矩阵是稀疏矩阵，稀疏矩阵只保存非零的元素，可以节省内存
# 稀疏矩阵：矩阵中，大多数值都为0的矩阵
dict_vec = DictVectorizer(sparse=False)

# 提取特征
X = dict_vec.fit_transform(data)

print("提取的特征：\n", X)

print("特征名称：\n", dict_vec.get_feature_names_out())

# 结合DataFrame查看结果
pd.DataFrame(X, columns=dict_vec.get_feature_names_out())
```

3.2 从英文文本加载特征

```python
# 导包
from sklearn.feature_extraction.text import CountVectorizer

# 样本文本
documents = [
    "I love machine learning. Machine learning is interesting.",
    "I love coding. Coding is fun and useful.",
    "Machine learning and coding are my favorite skills."
]

# 创建CountVectorizer对象
# min_df/max_df参数解释：
# min_df: 最小文档频率，即单词至少出现的次数或比例，低于此阈值的单词将被忽略
# max_df: 最大文档频率，即单词在文档中的次数或比例，高于此阈值的单词将被忽略
# 当min_df和max_df都为整数时，表示单词出现的次数，当min_df和max_df都为小数时，表示单词在文档中的比例
count_vec = CountVectorizer(min_df=1, max_df=2)

# 提取特征
X = count_vec.fit_transform(documents)

print("提取的特征：\n", X)

print("特征名称：\n", count_vec.get_feature_names_out())

# 结合DataFrame查看结果
df = pd.DataFrame(X.toarray(), columns=count_vec.get_feature_names_out())

print(df)

```

3.3 从中文文本加载特征

```python
# 导入分词
import jieba

# 文本
documents = [
    "人生苦短，我喜欢 python",
    "人生漫长，不用 python python",
    "人生漫漫，不用 python python python"
]

# 文本分词，得到一个分词之后的文本列表
# 使用jieba分词，把每个句子分词，然后通过空格再合并成一个句子
# CountVectorizer()是根据空格分词的，所以需要添加空格
documents_cut = [' '.join(jieba.cut(document)) for document in documents]

# 创建CountVectorizer对象
count_vec = CountVectorizer()

# 提取特征
X = count_vec.fit_transform(documents_cut)

# 使用DataFrame查看结果
pd.DataFrame(X.toarray(), columns=count_vec.get_feature_names_out())
```

3.4 TF-IDFVectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# 样本文本
documents = [
    "I love machine learning. Machine learning is interesting.",
    "I love coding. Coding is fun and useful.",
    "Machine learning and coding are my favorite skills."
]

# 创建TF-IDF对象
# stopwords='english' 忽略英文中的停用词
# 停用词：指文本中频繁出现，但通常对语义理解帮助不大的虚词或常见词
tfidf_vec = TfidfVectorizer(min_df=1, stop_words='english')

# 创建特征矩阵
x = tfidf_vec.fit_transform(documents)

# 结合DataFrame查看结果
pd.DataFrame(x.toarray(), columns=tfidf_vec.get_feature_names_out())
```
