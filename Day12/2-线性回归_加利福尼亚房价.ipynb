```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

# 加载加利福尼亚房价数据集
california = fetch_california_housing()

# 显示基本信息
print("数据集描述信息：")
print(california.DESCR[:500])  # 只显示前500字符简介
print("数据形状：", california.data.shape)
print("特征名：", california.feature_names)
"""
MedInc：街区住户的收入中位数（单位是“万美金”量级的缩放值，越大一般房价越高）
HouseAge：街区房屋的平均房龄（年）
AveRooms：每户的平均房间数（总房间数 / 总户数）
AveBedrms：每户的平均卧室数（总卧室数 / 总户数）
Population：街区的人口数
AveOccup：每户的平均居住人数（人口数 / 总户数）
Latitude：街区的纬度
Longitude：街区的经度
"""
print("目标名：", california.target_names)
"""
MedHouseVal：街区房价的中位数（通常以“10 万美元”为单位的缩放值；例如 2.5 表示约 25 万美元）
"""

# 查看数据集（转为DataFrame方便查看）
df = pd.DataFrame(california.data, columns=california.feature_names)
df['MedHouseVal'] = california.target

print("\n数据集前5行：")
print(df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 划分特征和目标
X = california.data
y = california.target

# 划分数据集，训练集比例为0.2
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.2, random_state=42)

# 标准化数据
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 线性回归模型（正规方程实现）
lin_reg = LinearRegression()
lin_reg.fit(X_train_scaled, y_train)

# 对测试集进行预测
y_test_pred = lin_reg.predict(X_test_scaled)

# 输出测试集预测前5个值
print("测试集预测前5个值：", y_test_pred[:5])

# 计算均方误差和R^2分数
mse = mean_squared_error(y_test, y_test_pred)
r2 = r2_score(y_test, y_test_pred)
print("测试集均方误差 (MSE)：", mse)
print("测试集R^2分数：", r2)


```

```python
from sklearn.linear_model import SGDRegressor

# 使用sklearn的SGDRegressor（随机梯度下降线性回归）
sgd_reg = SGDRegressor(max_iter=2000, learning_rate='invscaling', eta0=0.01, random_state=42)
# 参数含义说明
# max_iter：最大迭代次数。设置为2000表示最多迭代2000次以收敛。
# learning_rate：学习率调整策略。'invscaling' 表示学习率随迭代次数递减（η = eta0 / pow(t, power_t)）。
# eta0：初始学习率。这里设为0.01。
# random_state：随机数种子，保证结果可复现。

# 模型训练
sgd_reg.fit(X_train_scaled, y_train)

# 对测试集进行预测
y_test_pred_sgd = sgd_reg.predict(X_test_scaled)

print("SGD梯度下降测试集预测前5个值：", y_test_pred_sgd[:5])

# 计算均方误差和R^2分数
mse_sgd = mean_squared_error(y_test, y_test_pred_sgd)
r2_sgd = r2_score(y_test, y_test_pred_sgd)
print("SGD梯度下降测试集均方误差 MSE：", mse_sgd)
print("SGD梯度下降测试集R^2分数：", r2_sgd)

```

### 调整学习率观察结果

```python
# 使用sklearn的SGDRegressor（随机梯度下降线性回归）
sgd_reg = SGDRegressor(max_iter=2000, learning_rate='invscaling', eta0=0.001, random_state=42)

# 模型训练
sgd_reg.fit(X_train_scaled, y_train)

# 对测试集进行预测
y_test_pred_sgd = sgd_reg.predict(X_test_scaled)

print("SGD梯度下降测试集预测前5个值：", y_test_pred_sgd[:5])

# 计算均方误差和R^2分数
mse_sgd = mean_squared_error(y_test, y_test_pred_sgd)
r2_sgd = r2_score(y_test, y_test_pred_sgd)
print("SGD梯度下降测试集均方误差 MSE：", mse_sgd)
print("SGD梯度下降测试集R^2分数：", r2_sgd)
```

**学习率 eta0=0.001 的梯度下降预测结果更好**

**原因分析（为什么学习率一大就“炸”）**

学习率太大（0.01）= 每步走得太猛：即使你做了标准化，SGDRegressor 仍是“随机梯度”（梯度带噪声），步子一大就容易在最优点附近反复越界，甚至越走越远，导致权重变得极大，预测值也就爆掉，最终 R^2 变成巨大负数。

invscaling 会衰减学习率，但“起步那一下”仍可能过大：前期更新就可能把参数推到不合理区域，后面再衰减也救不回来。

学习率更小（0.001）更稳：更新幅度更可控，能逐步逼近较优解，所以指标正常。
