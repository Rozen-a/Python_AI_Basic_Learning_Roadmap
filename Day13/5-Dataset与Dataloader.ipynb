## Dataset

```python
# 自定义Dataset
import torch
from torch.utils.data import Dataset

class MyDataset(Dataset):
    def __init__(self):
        # 1. 初始化数据，例如读取文件
        self.data = ['李云龙','楚云飞','赵刚','丁伟','孔捷','常乃超','魏大勇','秀芹']
        self.total = len(self.data)

    def __getitem__(self, index):
        # 2. 获取索引对应的数据
        return self.data[index]


    def __len__(self):
        # 3. 返回数据集的长度
        return self.total

if __name__ == '__main__':
    dataset = MyDataset()
    print(dataset[1])
```

## DataLoader

```python
# DataLoader的创建
import torch
from torch.utils.data import DataLoader

# 创建DataLoader
# batch_size: 每次批次的大小（一般是2的倍数）
# shuffle: 是否打乱数据
dataloader = DataLoader(MyDataset(), batch_size=2, shuffle=False)
	# 每次取两个样本（Tensor）装箱，打包成一个data（Tensor）
    # 此处示例使用的是列表而非Tensor，所以返回data的类型也是list而不是Tensor

# 获取批次数据
for data in dataloader:
    print(data)
```

```python
# 打乱示例
dataloader = DataLoader(MyDataset(), batch_size=2, shuffle=True)

# 获取批次数据
for data in dataloader:
    print(data)
```
